```python
# Import necessary libraries
import pandas as pd
from scipy import stats
import numpy as np
```

```python
# Load your results file — adjust path and column names to match yours
df = pd.read_csv("comparison_report.csv")
```

## Paired T-Tests

```python
# Filter to the service level column per agent
rl_lstm = df[df["Agent"] == "RL+LSTM Agent"]["Service Level (%)"].values
standard_rl = df[df["Agent"] == "Standard RL Agent"]["Service Level (%)"].values
fixed_s = df[df["Agent"] == "Fixed-S Policy"]["Service Level (%)"].values
base_policy = df[df["Agent"] == "Forecast Base-Stock"]["Service Level (%)"].values
```

```python
# Perform paired t-test between RL+LSTM and Base-Stock
t_stat, p_value = stats.ttest_rel(rl_lstm, base_policy)
print(f"t = {t_stat:.3f}, p = {p_value:.4f}")
```

```python
# Perform paired t-test between RL+LSTM and Standard RL
t_stat, p_value = stats.ttest_rel(rl_lstm, standard_rl)
print(f"t = {t_stat:.3f}, p = {p_value:.4f}")
```

```python
# Perform paired t-test between RL+LSTM and Fixed S Policy
t_stat, p_value = stats.ttest_rel(rl_lstm, fixed_s)
print(f"t = {t_stat:.3f}, p = {p_value:.4f}")
```

## Cohen d's Test

```python
def cohens_d(a, b):
    return (np.mean(a) - np.mean(b)) / (((np.std(a, ddof=1)**2 + np.std(b, ddof=1)**2) / 2) ** 0.5)

print(cohens_d(rl_lstm, standard_rl))
print(cohens_d(rl_lstm, base_policy))
print(cohens_d(rl_lstm, fixed_s))
```

```python
def bootstrap_ci(data, n_boot=10000, ci=95):
    means = [np.mean(np.random.choice(data, len(data), replace=True)) 
             for _ in range(n_boot)]
    lower = np.percentile(means, (100-ci)/2)
    upper = np.percentile(means, 100-(100-ci)/2)
    return lower, upper

lower, upper = bootstrap_ci(rl_lstm)
print(f"95% CI: [{lower:.2f}, {upper:.2f}]")
```
